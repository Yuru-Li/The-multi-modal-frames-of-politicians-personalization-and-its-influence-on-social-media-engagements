```python
# 处理缺失值，这里假设使用0进行填充
df.fillna(0, inplace=True)
```

```python
# 标准化连续变量
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df[continuous_columns])
```

```python
df_scaled
```

```python
# Convert all values in the categorical columns to strings
df[categorical_columns] = df[categorical_columns].astype(str)

# Apply one-hot encoding
encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')
df_encoded = pd.DataFrame(encoder.fit_transform(df[categorical_columns]))

# Assign appropriate column names to the encoded DataFrame
df_encoded.columns = encoder.get_feature_names_out(categorical_columns)
```

```python
# 合并标准化的连续变量和独热编码的定类变量
df_features = pd.DataFrame(df_scaled, columns=continuous_columns)
df_features = pd.concat([df_features, df_encoded], axis=1)
```

```python
# 进行主成分分析降维
pca = PCA(n_components=3)
df_pca = pca.fit_transform(df_features)
```

```python
# 聚类数量
num_clusters = 5

# 进行K均值聚类
kmeans = KMeans(n_clusters=num_clusters, random_state=42)
kmeans.fit(df_pca)

# 获取聚类标签和中心点
labels = kmeans.labels_
df['cluster_labels']=labels
centers = kmeans.cluster_centers_
```

```python
# 绘制聚类散点图
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=labels, cmap='viridis')
plt.scatter(centers[:, 0], centers[:, 1], c='red', marker='x')  # 绘制聚类中心点
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Fig.5. Cluster Distribution')
# 保存图像到本地
plt.savefig('figures/cluster_distribution.jpg', dpi=300)  # 保存为300 DPI的PNG格式
plt.show()
```

```python
# 轮廓系数
silhouette_avg = silhouette_score(df_pca, labels) #越接近1效果越好
print("轮廓系数:", silhouette_avg)
```

```python
print("Inertia:", kmeans.inertia_)
```

```python
# 获取簇中离中心最近的样本
cluster_samples = [[] for _ in range(num_clusters)]
distances = kmeans.transform(df_pca)  # Distances to each cluster centroid
k = 20  # Number of core samples to extract

for i in range(df_pca.shape[0]):
    cluster_label = labels[i]
    sample_distance = distances[i, cluster_label]
    cluster_samples[cluster_label].append((sample_distance, i))

core_samples = []
for cluster_label in range(num_clusters):
    samples = sorted(cluster_samples[cluster_label])[:k]
    core_samples.extend(samples)

# Save the core samples to a CSV file
core_samples_indices = [sample_idx for _, sample_idx in core_samples]
core_samples_df = df.iloc[core_samples_indices]
core_samples_df.to_csv('results/core_samples35.csv', index=False)
```

```python
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols

# 将 Likes 和 Comments 列转换为数值类型，无法转换的值将变为 NaN
df['Likes'] = pd.to_numeric(df['Likes'], errors='coerce')
df['Comments'] = pd.to_numeric(df['Comments'], errors='coerce')

# 删除包含 NaN 的行
df = df.dropna(subset=['Likes', 'Comments', 'cluster_labels'])

# ANOVA 检验 - Likes
model_likes = ols('Likes ~ C(cluster_labels)', data=df).fit()
anova_table_likes = sm.stats.anova_lm(model_likes, typ=2)
print("ANOVA Results for Likes:")
print(anova_table_likes)

# ANOVA 检验 - Comments
model_comments = ols('Comments ~ C(cluster_labels)', data=df).fit()
anova_table_comments = sm.stats.anova_lm(model_comments, typ=2)
print("\nANOVA Results for Comments:")
print(anova_table_comments)

# 如果数据不满足ANOVA假设，可以使用Kruskal-Wallis检验作为非参数检验
# Kruskal-Wallis 检验 - Likes
likes_groups = [df['Likes'][df['cluster_labels'] == label] for label in df['cluster_labels'].unique()]
kw_likes_stat, kw_likes_pval = stats.kruskal(*likes_groups)
print("\nKruskal-Wallis Test for Likes:")
print(f"Statistic: {kw_likes_stat}, p-value: {kw_likes_pval}")

# Kruskal-Wallis 检验 - Comments
comments_groups = [df['Comments'][df['cluster_labels'] == label] for label in df['cluster_labels'].unique()]
kw_comments_stat, kw_comments_pval = stats.kruskal(*comments_groups)
print("\nKruskal-Wallis Test for Comments:")
print(f"Statistic: {kw_comments_stat}, p-value: {kw_comments_pval}")
```

## age difference

```python
###############################################
# 2. 连续变量与 Age 的关系 (Pearson相关)
###############################################
pearson_age_results = []
for var in continuous_vars:
    valid_data = df[['Age', var]].dropna()
    n = len(valid_data)
    if n > 1:
        corr, p_val = stats.pearsonr(valid_data['Age'], valid_data[var])
        pearson_age_results.append([var, n, corr, p_val])
    else:
        pearson_age_results.append([var, n, None, None])

pearson_age_df = pd.DataFrame(pearson_age_results, columns=[
    "Variable", "N", "Pearson_Correlation", "p_value"
])
pearson_age_df.to_csv(os.path.join(results_dir, "continuous_vars_by_age_pearson.csv"), index=False)
```

```python
###############################################
# 4. 分类变量在 Age 上的差异 (ANOVA或t检验)
###############################################
age_cat_results = []
for cat_var in categorical_vars:
    categories = df[cat_var].dropna().unique()
    if len(categories) > 2:
        # ANOVA
        groups_data = [df.loc[df[cat_var] == cat, 'Age'].dropna() for cat in categories]
        if all([len(g) > 1 for g in groups_data]):
            f_stat, p_val = stats.f_oneway(*groups_data)
            counts = [len(g) for g in groups_data]
            age_cat_results.append([cat_var, "ANOVA", len(categories), f_stat, p_val, counts])
        else:
            age_cat_results.append([cat_var, "ANOVA", len(categories), None, None, None])
    elif len(categories) == 2:
        # t检验
        cat1, cat2 = categories
        g1 = df.loc[df[cat_var] == cat1, 'Age'].dropna()
        g2 = df.loc[df[cat_var] == cat2, 'Age'].dropna()
        if len(g1) > 1 and len(g2) > 1:
            t_stat, p_val = stats.ttest_ind(g1, g2, equal_var=False)
            age_cat_results.append([cat_var, "t-test", 2, t_stat, p_val, [len(g1), len(g2)]])
        else:
            age_cat_results.append([cat_var, "t-test", 2, None, None, [len(g1), len(g2)]])
    else:
        # 只有一个类别
        age_cat_results.append([cat_var, "No test", len(categories), None, None, None])

age_cat_df = pd.DataFrame(age_cat_results, columns=[
    "Categorical_Variable", "Test_Type", "Number_of_Categories", "Statistic", "p_value", "Group_sizes"
])
age_cat_df.to_csv(os.path.join(results_dir, "categorical_vars_by_age.csv"), index=False)
```

```python
import pandas as pd
import numpy as np
import os
from scipy import stats

# 假设df已存在，并包含以下列:
# 'Age', 'Gender', 'party'(7类), 'text_topic', 'img_topic', 'cluster_labels' 以及 continuous_vars
continuous_vars = [
    'text_PARTY_count', 'text_POLITICIAN_count', 'text_frames', 'text_negative', 
    'text_neutral', 'text_positive', 'img_party', 'img_Number of Faces', 
    'img_Average Face Size', 'img_face match', 'img_frames', 'img_cloths', 
    'img_negative', 'img_neutral', 'img_positive'
]

categorical_vars = ['text_topic_names', 'img_topic', 'cluster_labels']

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
for var in continuous_vars:
    df[var] = pd.to_numeric(df[var], errors='coerce')

results_dir = "analysis_spss_like_tables"
if not os.path.exists(results_dir):
    os.makedirs(results_dir)


###############################################
# 1. 连续变量在 Gender 上的差异 (t检验)
###############################################
gender_groups = df['Gender'].dropna().unique()
ttest_gender_results = []
if len(gender_groups) == 2:
    g1_label, g2_label = gender_groups
    group1_data_full = df[df['Gender'] == g1_label]
    group2_data_full = df[df['Gender'] == g2_label]

    for var in continuous_vars:
        g1 = group1_data_full[var].dropna()
        g2 = group2_data_full[var].dropna()
        N1, N2 = len(g1), len(g2)
        mean1, mean2 = g1.mean() if N1>0 else np.nan, g2.mean() if N2>0 else np.nan
        sd1, sd2 = g1.std(ddof=1) if N1>1 else np.nan, g2.std(ddof=1) if N2>1 else np.nan
        
        if N1 > 1 and N2 > 1:
            t_stat, p_val = stats.ttest_ind(g1, g2, equal_var=False)
            # 自由度近似值
            # Welch's t-test自由度统计可以使用scipy默认结果，也可自己计算
            # 这里使用一种近似计算方法
            df_num = (sd1**2/N1 + sd2**2/N2)**2
            df_den = (sd1**4/(N1**2*(N1-1))) + (sd2**4/(N2**2*(N2-1)))
            df_val = df_num / df_den if df_den != 0 else np.nan
        else:
            t_stat, p_val, df_val = np.nan, np.nan, np.nan

        ttest_gender_results.append([
            var, N1, mean1, sd1, N2, mean2, sd2, t_stat, df_val, p_val
        ])

ttest_gender_df = pd.DataFrame(ttest_gender_results, columns=[
    "Variable", "N(Group1)", "Mean(Group1)", "SD(Group1)", 
    "N(Group2)", "Mean(Group2)", "SD(Group2)",
    "t", "df", "Sig(2-tailed)"
])
ttest_gender_df.to_csv(os.path.join(results_dir, "gender_ttest_results.csv"), index=False)


###############################################
# 2. 连续变量与 Age 的相关 (Pearson)
###############################################
pearson_age_results = []
for var in continuous_vars:
    valid_data = df[['Age', var]].dropna()
    N = len(valid_data)
    if N > 1:
        r, p_val = stats.pearsonr(valid_data['Age'], valid_data[var])
    else:
        r, p_val = np.nan, np.nan
    pearson_age_results.append([var, N, r, p_val])

pearson_age_df = pd.DataFrame(pearson_age_results, columns=[
    "Variable", "N", "Pearson r", "Sig(2-tailed)"
])
pearson_age_df.to_csv(os.path.join(results_dir, "age_pearson_results.csv"), index=False)


###############################################
# 3. 分类变量与 Gender 的关系 (卡方检验)
###############################################
chi_gender_results = []
for cat_var in categorical_vars:
    contingency_table = pd.crosstab(df['Gender'], df[cat_var])
    if contingency_table.size > 0 and contingency_table.shape[0] > 1 and contingency_table.shape[1] > 1:
        chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)
    else:
        chi2, p_val, dof = np.nan, np.nan, np.nan

    # 将频数表展开
    freq_records = []
    for row_cat in contingency_table.index:
        for col_cat in contingency_table.columns:
            freq_records.append([cat_var, row_cat, col_cat, contingency_table.loc[row_cat, col_cat]])
    freq_df = pd.DataFrame(freq_records, columns=["Variable", "RowCategory", "ColumnCategory", "Count"])
    freq_df['Variable'].iloc[0] = f"{cat_var}(Chi-square: {chi2:.2f}, p={p_val:.3f}, df={dof})" if not pd.isnull(chi2) else cat_var
    freq_df.to_csv(os.path.join(results_dir, f"{cat_var}_gender_chi2_results.csv"), index=False)


###############################################
# 4. 分类变量在 Age 上的差异 (ANOVA或t检验)
###############################################
age_cat_results = []
for cat_var in categorical_vars:
    categories = df[cat_var].dropna().unique()
    if len(categories) > 2:
        # ANOVA
        groups_data = [df.loc[df[cat_var] == cat, 'Age'].dropna() for cat in categories]
        if all([len(g) > 1 for g in groups_data]):
            f_stat, p_val = stats.f_oneway(*groups_data)
            # df 计算
            df_between = len(categories)-1
            # 总df = N-1
            N_total = sum([len(g) for g in groups_data])
            df_within = N_total - len(categories)
        else:
            f_stat, p_val, df_between, df_within = np.nan, np.nan, np.nan, np.nan
        # 输出描述统计
        desc_records = []
        for cat in categories:
            d = df.loc[df[cat_var] == cat, 'Age'].dropna()
            desc_records.append([cat_var, cat, len(d), d.mean() if len(d)>0 else np.nan, d.std(ddof=1) if len(d)>1 else np.nan])
        desc_df = pd.DataFrame(desc_records, columns=["Variable", "Category", "N", "Mean", "SD"])
        # 添加ANOVA结果行
        desc_df.loc[len(desc_df)] = [f"ANOVA for {cat_var}", "", "", f"F={f_stat:.2f}", f"p={p_val:.3f}"]
        desc_df.to_csv(os.path.join(results_dir, f"{cat_var}_age_anova_results.csv"), index=False)

    elif len(categories) == 2:
        # t检验
        cat1, cat2 = categories
        g1 = df.loc[df[cat_var] == cat1, 'Age'].dropna()
        g2 = df.loc[df[cat_var] == cat2, 'Age'].dropna()
        N1, N2 = len(g1), len(g2)
        mean1, mean2 = g1.mean() if N1>0 else np.nan, g2.mean() if N2>0 else np.nan
        sd1, sd2 = g1.std(ddof=1) if N1>1 else np.nan, g2.std(ddof=1) if N2>1 else np.nan
        if N1>1 and N2>1:
            t_stat, p_val = stats.ttest_ind(g1, g2, equal_var=False)
            # Welch df
            df_num = (sd1**2/N1 + sd2**2/N2)**2
            df_den = (sd1**4/(N1**2*(N1-1))) + (sd2**4/(N2**2*(N2-1)))
            df_val = df_num/df_den if df_den>0 else np.nan
        else:
            t_stat, p_val, df_val = np.nan, np.nan, np.nan

        ttest_age_df = pd.DataFrame([[
            cat_var, cat1, N1, mean1, sd1,
            cat2, N2, mean2, sd2,
            t_stat, df_val, p_val
        ]], columns=[
            "Variable", "Group1", "N(Group1)", "Mean(Group1)", "SD(Group1)",
            "Group2", "N(Group2)", "Mean(Group2)", "SD(Group2)",
            "t", "df", "Sig(2-tailed)"
        ])
        ttest_age_df.to_csv(os.path.join(results_dir, f"{cat_var}_age_ttest_results.csv"), index=False)


###############################################
# 5. party有7个类别，对连续变量用ANOVA，以及Age用ANOVA
###############################################
party_categories = df['party'].dropna().unique()
if len(party_categories) > 2:
    # 连续变量在party下差异(ANOVA)
    for var in continuous_vars:
        groups_data = [df.loc[df['party'] == cat, var].dropna() for cat in party_categories]
        if all(len(g)>1 for g in groups_data):
            f_stat, p_val = stats.f_oneway(*groups_data)
            df_between = len(party_categories)-1
            N_total = sum([len(g) for g in groups_data])
            df_within = N_total - len(party_categories)
        else:
            f_stat, p_val, df_between, df_within = np.nan, np.nan, np.nan, np.nan
        # 描述统计
        desc_records = []
        for cat in party_categories:
            d = df.loc[df['party'] == cat, var].dropna()
            desc_records.append([var, cat, len(d), d.mean() if len(d)>0 else np.nan, d.std(ddof=1) if len(d)>1 else np.nan])
        desc_df = pd.DataFrame(desc_records, columns=["Variable", "Category", "N", "Mean", "SD"])
        desc_df.loc[len(desc_df)] = [f"ANOVA for {var}", "", "", f"F={f_stat:.2f}", f"p={p_val:.3f}"]
        desc_df.to_csv(os.path.join(results_dir, f"{var}_party_anova_results.csv"), index=False)

    # 分类变量与party的卡方
    for cat_var in categorical_vars:
        contingency_table = pd.crosstab(df['party'], df[cat_var])
        if contingency_table.shape[0]>1 and contingency_table.shape[1]>1:
            chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)
        else:
            chi2, p_val, dof = np.nan, np.nan, np.nan
        # 输出列联表带统计结果
        freq_records = []
        for row_cat in contingency_table.index:
            for col_cat in contingency_table.columns:
                freq_records.append([cat_var, row_cat, col_cat, contingency_table.loc[row_cat, col_cat]])
        freq_df = pd.DataFrame(freq_records, columns=["Variable", "RowCategory", "ColumnCategory", "Count"])
        freq_df.loc[len(freq_df)] = [f"Chi-square for {cat_var}", "", "", f"Chi2={chi2:.2f}, p={p_val:.3f}, df={dof}"]
        freq_df.to_csv(os.path.join(results_dir, f"{cat_var}_party_chi2_results.csv"), index=False)

    # Age在party下(ANOVA)
    groups_data = [df.loc[df['party'] == cat, 'Age'].dropna() for cat in party_categories]
    if all(len(g)>1 for g in groups_data):
        f_stat, p_val = stats.f_oneway(*groups_data)
    else:
        f_stat, p_val = np.nan, np.nan
    desc_records = []
    for cat in party_categories:
        d = df.loc[df['party'] == cat, 'Age'].dropna()
        desc_records.append(['Age', cat, len(d), d.mean() if len(d)>0 else np.nan, d.std(ddof=1) if len(d)>1 else np.nan])
    desc_df = pd.DataFrame(desc_records, columns=["Variable", "Category", "N", "Mean", "SD"])
    desc_df.loc[len(desc_df)] = [f"ANOVA for Age by party", "", "", f"F={f_stat:.2f}", f"p={p_val:.3f}"]
    desc_df.to_csv(os.path.join(results_dir, "Age_party_anova_results.csv"), index=False)


print("SPSS风格的统计表格已输出至:", results_dir)
```
