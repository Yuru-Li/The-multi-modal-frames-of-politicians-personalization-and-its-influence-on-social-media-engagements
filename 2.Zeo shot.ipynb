```python
#import re
import os
import pandas as pd
#import numpy as np
import glob
import warnings
import torch
from PIL import Image
from transformers import CLIPProcessor, CLIPModel
```

```python
# 指定要使用的GPU设备编号，通常为0号GPU
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

# Check if a GPU is available
if torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")

print("Running on:", device)
```

```python
# Load CLIP model
model2 = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
model2.to(device)
model2 = model2.eval()
# Load CLIP processor
processor2 = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
global processor2
global model2
```

def extract_filename(path):
    # 从路径中提取文件名，仅保留 "processed image 100 examples/" 后面的数据
    prefix = "post images/"
    if prefix in path:
        return path.split(prefix, 1)[1]

```python
def poster(img):
    #image = Image.open(image_path)
    text=['poster', 'not poster']
    inputs=processor2(text=text, images=image, return_tensors='pt', padding=True)
    inputs = {key: value.to(device) for key, value in inputs.items()}
    outputs=model2(**inputs)
    logits_per_image=outputs.logits_per_image
    probs=logits_per_image.softmax(dim=1)
    max_prob, max_index = torch.max(probs[0], dim=0)
    poster = text[max_index]

    if poster == 'not poster':
        return None  # Return missing value if the result is 'not poster'
    else:
        return poster
```
